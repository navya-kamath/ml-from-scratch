```python
import numpy as np
```

```python
a = [[1,2],[3,4]] #Square matrix
b = [[1,2,3],[3,4,5]] #Rectangular matrix
```

```python
#Transpose of square and rectangular matrix
def transpose(matrix):
    transpose_matrix = []
    for i in range(len(matrix[0])):
        row = []
        for j in range(len(matrix)):
            row.append(matrix[j][i])
        transpose_matrix.append(row)
    return transpose_matrix       
```

```python
print(transpose(a))
```

```python
print(transpose(b))
```

```python
#Tranpose using numpy
np.transpose(a)
```

```python
np.transpose(b)
```

```python
#Is Symmetric
s = [[1,2],[2,1]]

def is_symmetric(matrix):
    transpose_matrix = []
    for i in range(len(matrix[0])):
        row = []
        for j in range(len(matrix)):
            row.append(matrix[j][i])
        transpose_matrix.append(row)
    if matrix == transpose_matrix:
        return "The matrix is symmetric"
    else:
        return "The matrix is not symmetrix"
```

```python
print(is_symmetric(s))
```

```python
print(is_symmetric(a))
```

```python
#A'A in multiply_matrix check if its symmetric
def multiply_matrix(m1, m2):
    if len(m1[0]) != len(m2):  # Check if multiplication is possible
        return "Matrix multiplication not possible"

    prod_matrix = []
    for i in range(len(m1)):
        row = []
        for j in range(len(m2[0])):
            total = 0
            for k in range(len(m1[0])):
                total += m1[i][k] * m2[k][j]
            row.append(total)
        prod_matrix.append(row)

    return prod_matrix

def check_symmetry(m1):
    prod_sym = multiply_matrix(transpose(m1), m1)

    if is_symmetric(prod_sym):
        return "Product is symmetric"
    else:
        return "Product is not symmetric"

print(check_symmetry(a))
print(check_symmetry(s))
```
