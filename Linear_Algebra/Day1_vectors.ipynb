```python
import numpy as np
```

```python
#Vectors
v1 = [2,4,6]
v2 = [1,3,5]
print(f"Vector v1 is {v1}")
print(f"Vector v2 is {v2}")
```

```python
#Vector addition
result = []

for i in range(len(v1)):
    result.append(v1[i]+v2[i])
print(f"Vector addition v1 + v2 = {result}")
```

```python
#Vector addition using numpy
np.add(v1,v2)
```

```python
#Scalar multiplication of a vector
scalar_mul = []

for i in range(len(v1)):
    scalar_mul.append(3*v1[i])
print(f"Scalar multiplication of a vector {scalar_mul}") 
```

```python
#Scalar multiplication using numpy
np.multiply(3,v1)
```

```python
#Dot product of vector
dot_product = 0
vector_mul = []
for i in range(len(v1)):
    vector_mul.append(v1[i]*v2[i])
    dot_product += vector_mul[i]
print(f"Dot product of vector v1 and v2 is {dot_product}")
```

```python
#Dot product using numpy
np.dot(v1,v2)
```

```python
#Norm of a vector
squared_sum = 0
for i in range(len(v1)):
    squared_sum += v1[i]**2
    norm = squared_sum**(1/2)
print(f"Norm of vector v1 is {norm}")
```

```python
#Norm using numpy
np.linalg.norm(v1)
```

```python
#Verify whether the dimension of vector are same or not
def is_same_dimension(a,b):
    if np.shape(a) == np.shape(b):
        return True
    else:
        return False

print(is_same_dimension(v1,v2))
print(is_same_dimension([1,2],[3,4,5]))
```
