```python
#Swap Rows
a = [[1,2],[3,4]]
b = [[1,2,3],[3,4,5]]
c = [[1,2],[3,4],[5,6]]

def swap_rows(matrix, row1, row2):
    row_swap_mat = [row[:] for row in matrix]  # Make a copy
    row_swap_mat[row1], row_swap_mat[row2] = row_swap_mat[row2], row_swap_mat[row1]
    return row_swap_mat

print(swap_rows(a,0,1))
print(swap_rows(b,0,1))
print(swap_rows(c,0,2))
```

```python
#Create Permutation matrix
def permutation_matrix(n, row1, row2):
    identity_matrix = []

    for i in range(n):
        row = []
        for j in range(n):
            if i == j:
                row.append(1)
            else:
                row.append(0)
        identity_matrix.append(row)

    return swap_rows(identity_matrix, row1, row2)

print(permutation_matrix(3, 0, 1))
```

```python
#Create permutation using numpy
def permutation_matrix(n,row1,row2):
    identity = np.identity(n)
    return swap_rows(identity, row1, row2)

print(permutation_matrix(3,0,1))
```
