```python
import numpy as np
```

```python
#Vector
v = [1,2]

#Matrix
m = [[1,2,3],
     [2,4,6]]
```

```python
#Print a matrix which is 2 X 3
print(np.array(m))
```

```python
#Matric addition
m1 = [[1,2],[3,4]]
m2 = [[2,4],[4,5]]

matrix_add = []

for i in range(len(m1)):
    row = []                     # Create a new row
    for j in range(len(m1[0])):
        row.append(m1[i][j] + m2[i][j])   # Fill the row
    matrix_add.append(row)       # Add completed row to matrix
print(f"Matrix m1 + m2 = {matrix_add}")
```

```python
#Matrix addition using Numpy
np.add(m1,m2)
```

```python
#Matrix subtraction
matrix_subtract = []

for i in range(len(m1)):
    row = []                     # Create a new row
    for j in range(len(m1[0])):
        row.append(m1[i][j] - m2[i][j])   # Fill the row
    matrix_subtract.append(row)       # Add completed row to matrix

print(f"Matrix m1 - m2 = {matrix_subtract}")
```

```python
#Matrix subtraction using numpy
np.subtract(m1,m2)
```

```python
#Scalar and matrix multiplication
s = 3

scalar_matrix = []
for i in range(len(m1)):
    row= []
    for j in range(len(m1[0])):
        row.append(s* m1[i][j])
    scalar_matrix.append(row)
print(scalar_matrix)
```

```python
#Scalar and matrix multiplication using numpy
np.multiply(s,m1)
```

```python
#Matric and vector multiplication
v = [1,2]

vector_matrix = []
for i in range(len(m1)):
    row = []
    for j in range(len(v)):
        row.append(m1[i][j]*v[j])
    vector_matrix.append(row)

print(vector_matrix)
```

```python
#Matric and vector multiplication using numpy 
np.multiply(m1,v)
```
