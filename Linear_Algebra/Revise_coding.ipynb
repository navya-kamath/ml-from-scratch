## Vectors

```python
#add_vectors()
def add_vectors(v1,v2):
    sum_vector = []
    if len(v1) == len(v2):
        for i in range(len(v1)):
            sum_vector.append(v1[i]+v2[i])
        return sum_vector
    else: 
        return "Addition not possible"
```

```python
a = [1,2,3,4,5]
b = [9,8,7,6,5]

print(add_vectors(a,b))
```

```python
#vector addition using the numpy
np.add(a,b)
```

```python
def dot_product(v1,v2):
    vector_dot_prod = 0
    if len(v1) == len(v2):
        for i in range(len(v1)):
            vector_dot_prod += (v1[i]*v2[i])
        return vector_dot_prod
    else:
        return "Dot product is not possible"
```

```python
print(dot_product(a,b))
```

```python
#Dot product using numpy
np.dot(a,b)
```

```python
#norm()
def norm(v1):
    norm_of_vector = 0
    for i in v1:
        total = 0
        for j in range(len(v1)):
            total += (v1[j]**2)
    norm_of_vector += total**(1/2)
    return norm_of_vector
```

```python
print(norm(a))
```

```python
#norm using numpy
np.linalg.norm(a)
```
