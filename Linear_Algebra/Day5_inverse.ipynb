```python
#Check whether matrix is square
m = [[1,2],[3,4]]

def square_matrix_check(matrix):
    if len(matrix) == len(matrix[0]):
        return f"Matrix is a square matrix"
    else:
        return f"Matrix is not a square matrix"

check_square_matrix = square_matrix_check(m)
print(check_square_matrix)

m_not = [[1,2,3],[2,3,4]]
print(square_matrix_check(m_not))
```

```python
#Check whether matrix is square using numpy
def sq_matrix(mat):
    mat = np.array(mat)
    if mat.shape[0] == mat.shape[1]:
        return f"Matrix is a square matrix"
    else:
        return f"Matrix is not a square matrix"

print(sq_matrix(m))
print(sq_matrix(m_not))
```

```python
#Generate identity matrix
def identity_matrix(n):
    mat = []
    for i in range(n):
        row = []
        for j in range(n):
            if i == j:
                row.append(1)
            else:
                row.append(0)
        mat.append(row)
    return mat

print(identity_matrix(3))
print(identity_matrix(4))
```

```python
#Matrix multiplication with identity matrix
m = [[1,2],[3,4]]
iden = [[1,0],[0,1]]

def identity_x_mat(mat,identity):
    if len(mat) == len(identity[1]):
        matrix_mul = []
        for i in range(len(mat)):
            row = []
            for j in range(len(identity[0])):
                total = 0
                for k in range(len(mat[0])):
                    total += mat[i][k] * identity[k][j]
                row.append(total)
            matrix_mul.append(row)
    return matrix_mul

matrix_mul = identity_x_mat(m,iden)

def verify_mul(matrix_multiplied,mat):
    if matrix_multiplied == mat:
        return "Correct answer"
    else:
        return "Wrong coding"

print(matrix_mul)
print(verify_mul(matrix_mul,m))
```

```python
a = np.random.randint(10, size = (3,3))
b = np.identity(3)

def identity_matrix_mul(mat, identity):
    mat = np.array(mat)
    identity = np.array(identity)

    # Check if multiplication is possible
    if mat.shape[1] == identity.shape[0]:
        matrix_mul = []

        for i in range(len(mat)):
            row = []
            for j in range(len(identity[0])):
                total = 0
                for k in range(len(mat[0])):
                    total += int(mat[i][k]) * int(identity[k][j])
                row.append(total)
            matrix_mul.append(row)

        return matrix_mul
    else:
        return None


matrix_mul = identity_matrix_mul(a, b)


def verify_multiplication(matrix_multiplied, mat):
    if np.array_equal(matrix_multiplied, mat):
        return "Correct answer"
    else:
        return "Wrong coding"


print(matrix_mul)
print(verify_multiplication(matrix_mul, a))
```
