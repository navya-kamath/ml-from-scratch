```python
import numpy as np
```

```python
v = [1,2,3]
w = [2,4,6]
```

```python
#Length (norm) of a vector
result = 0
for i in range(len(v)):
    result += v[i]**2
    norm = result ** (1/2)
print(f"Norm of vector v is {norm}")
```

```python
#Length (norm) using numpy
np.linalg.norm(v)
```

```python
#Dot product of two vector
dot_product = 0
for i in range(len(v)):
    dot_product += v[i]*w[i]
print(dot_product)
```

```python
#Dot product using numpy
np.dot(v,w)
```

```python
#Function to check if 2 vectors are othogonal i.e., their dot product is equal to zero.
def orthogonal_check(a,b):
    dot_product = 0
    for i in range(len(a)):
        dot_product += a[i] * b[i]
        
    if dot_product == 0:
        return "Vectors are othogonal"
    else:
        return "Vectors are not orthogonal"

result = orthogonal_check(v,w)
print(result)
```

```python
#Example for orthogonal vectors
a = [3,4]
b = [-4,3]
print(orthogonal_check(a,b))
```

```python
#Orthogonality check using numpy
def orthogonality_check(a,b):
    if np.dot(a,b) == 0: 
        return "Vectors are orthogonal"
    else:
        return "Vectors are not orthogonal"

print(orthogonality_check(a,b))
```

```python
#Norm and Dot product functions
def norm_of_vector(a):
    result = 0
    for i in range(len(a)):
        result += a[i] **2
        norm = result ** (1/2)
    return norm

print(norm_of_vector(a))

def dot_product(a,b):
    product = 0
    for i in range(len(a)):
        product += a[i] * b[i]
    return product

print(dot_product(a,b))
```

```python
#Function find cosine similarity
x = [1,2]
y = [3,5]
def cosine_similarity(x,y):
    cos_theta = dot_product(x,y) / (norm_of_vector(x) * norm_of_vector(y))
    return cos_theta

print(cosine_similarity(x,y))
```
